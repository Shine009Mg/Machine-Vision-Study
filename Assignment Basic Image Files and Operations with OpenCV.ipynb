```python
## Shine Lin Htet 6730161035
import cv2
import numpy as np

cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Error: Cannot open camera")
    exit()


x1, y1 = 200, 150
x2, y2 = 500, 400

captured_frame = None

while True:
    ret, frame = cap.read()

    if not ret:
        print("Error: Cannot read frame")
        break

    cv2.rectangle(
        frame,
        (x1, y1),
        (x2, y2),
        (255, 0, 0),
        2
    )

    cv2.putText(
        frame,
        "ROI",
        (x1, y1 - 10),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.7,
        (255, 0, 0),
        2
    )


    cv2.imshow("Camera", frame)

    key = cv2.waitKey(1) & 0xFF


    if key == ord('c'):

        captured_frame = frame.copy()

        roi = captured_frame[y1:y2, x1:x2]

        hsv = cv2.cvtColor(roi, cv2.COLOR_BGR2HSV)

        # Red has two ranges in HSV
        lower_red1 = np.array([0, 100, 100])
        upper_red1 = np.array([10, 255, 255])

        lower_red2 = np.array([170, 100, 100])
        upper_red2 = np.array([180, 255, 255])

       
        mask1 = cv2.inRange(
            hsv,
            lower_red1,
            upper_red1
        )

        mask2 = cv2.inRange(
            hsv,
            lower_red2,
            upper_red2
        )

       
        red_mask = mask1 + mask2

       
        red_pixels = cv2.countNonZero(red_mask)

      
        total_pixels = roi.shape[0] * roi.shape[1]

        
        red_percentage = (red_pixels / total_pixels) * 100

        if red_percentage > 5:

            cv2.putText(
                captured_frame,
                "Red object found",
                (30, 50),
                cv2.FONT_HERSHEY_SIMPLEX,
                1,
                (0, 0, 255),
                2
            )

            print("Red object found")

        else:

            cv2.putText(
                captured_frame,
                "No red object",
                (30, 50),
                cv2.FONT_HERSHEY_SIMPLEX,
                1,
                (255, 255, 255),
                2
            )

            print("No red object")

        # Show captured frame
        cv2.imshow("Captured Image", captured_frame)


    elif key == ord('s'):

        if captured_frame is not None:

            cv2.imwrite(
                "captured_roi.jpg",
                captured_frame[y1:y2, x1:x2]
            )

            print("Image saved as captured_image.jpg")

        else:
            print("No image captured. Press 'c' first.")


    elif key == ord('q'):
        break



cap.release()
cv2.destroyAllWindows()
```
